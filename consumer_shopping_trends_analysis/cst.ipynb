```python
from sklearn.preprocessing import LabelEncoder
```

```python
temp_df = df.copy()

for col in temp_df.select_dtypes(include='object').columns:
    temp_df[col] = LabelEncoder().fit_transform(temp_df[col])

plt.figure(figsize=(16,10))
sns.heatmap(temp_df.corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Age Group
df['age_group'] = pd.cut(df['age'],
                         bins=[18, 25, 35, 50, 65],
                         labels=['Youth', 'Young Adult', 'Adult', 'Senior'])

# High Income Customer
df['high_income'] = np.where(df['monthly_income'] > df['monthly_income'].median(), 1, 0)

# Frequent Online Shopper
df['frequent_online_shopper'] = np.where(df['monthly_online_orders'] > df['monthly_online_orders'].median(), 1, 0)

# Frequent Store Visitor
df['frequent_store_visitor'] = np.where(df['monthly_store_visits'] > df['monthly_store_visits'].median(), 1, 0)

df.head()
```

```python
X = df.drop('shopping_preference', axis=1)
y = df['shopping_preference']
```

```python
X_encoded = X.copy()

# Encode categorical columns
for col in X_encoded.select_dtypes(include='object').columns:
    X_encoded[col] = LabelEncoder().fit_transform(X_encoded[col])

# Encode age_group separately
X_encoded['age_group'] = LabelEncoder().fit_transform(X_encoded['age_group'])

# Encode target
y_encoded = LabelEncoder().fit_transform(y)

X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from xgboost import XGBClassifier
```

LOGISTIC REGRESSION

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_scaled, y_train)

y_pred_lr = lr.predict(X_test_scaled)

print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_lr))
print("\nClassification Report:\n", classification_report(y_test, y_pred_lr))
```

RANDOM FOREST CLASSIFIER

```python
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))
```

XGBOOST CLASSIFIER

```python
xgb = XGBClassifier(random_state=42, eval_metric='mlogloss')
xgb.fit(X_train, y_train)

y_pred_xgb = xgb.predict(X_test)

print("XGBoost Accuracy:", accuracy_score(y_test, y_pred_xgb))
print("\nClassification Report:\n", classification_report(y_test, y_pred_xgb))
```

```python
results = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest', 'XGBoost'],
    'Accuracy': [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_xgb),
    ]
})

results.sort_values(by='Accuracy', ascending=False)
```
